```python
cap = cv2.VideoCapture('Introduction.mp4')
if not cap.isOpened():
    print("Cannot open camera")


print("Press 'c' to capture face, 'q' to quit")

captured = False
while True:
    ret, frame = cap.read()
    if not ret:
        print("Failed to capture frame")
        break

    cv2.imshow('Registration', frame)
    key = cv2.waitKey(1) & 0xFF
    if key == ord('c'):
        captured = True
        break
    elif key == ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
```

```python
# image edge detection using canny
img = cv2.imread('goku.jpg')
img=cv2.resize(img, (600, 600))
cv2.imshow('Original Image', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
edges = cv2.Canny(gray, 100, 200)
cv2.imshow('Canny Edges', edges)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Video edge detection using canny
cap = cv2.VideoCapture(0)
if not cap.isOpened():
    print("Cannot open camera")
while True:
    ret, frame = cap.read()
    if not ret:
        print("Failed to capture frame")
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    edges = cv2.Canny(gray, 100, 200)

    cv2.imshow('Canny Edges', edges)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
```

```python
import numpy as np
kernel = np.ones((1,1), np.uint8)
img = cv2.imread('goku.jpg')
img=cv2.resize(img, (400, 400))
```

```python
# image eroison
gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
edges = cv2.Canny(gray, 400, 400)

erode_img = cv2.erode(edges, kernel, iterations=1)
# cv2.imshow('Canny Edges', erode_img)
# cv2.waitKey(0)
# cv2.destroyAllWindows()
```

```python
dil_img = cv2.dilate(erode_img, kernel, iterations=1)
cv2.imshow('Dilated Image', dil_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
